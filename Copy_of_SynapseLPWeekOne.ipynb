```python
# Write a class in Python for a point (1D)
# Use functions like __str__ for it. Find out what they do and write down some other dunder functions and what they do.

class Point1D:
    def __init__(self, x):
        self.x = x
#is a constructor and is automatically called when a new instance of class of created

    def __str__(self):
        return f"Point at x = {self.x}"
#defines what will be printed in the print() and is informal string representation


    def __repr__(self):
        return f"Point1D({self.x})"
#is the formal string representation

    def __eq__(self, other):
        return self.x == other.x
#defines the = function

    def __add__(self, other):
        return Point1D(self.x + other.x)
#defines the + function

    def __sub__(self, other):
        return Point1D(self.x - other.x)
#defines the - function
    def __len__(self):
#defines the len() of a point
        return abs(self.x)
    def __getitem__(self, key):
        if key == 0:
            return self.x
        else:
            raise IndexError("Point1D only supports index 0")
#defines indexing indexing of objects
p1 = Point1D(10)
p2 = Point1D(-3)

print(p1)
print(repr(p2))

print(p1 + p2)
print(p1 - p2)

print(p1 == p2)
print(len(p2))
print(p1[0])
#print(p1[1])
# Extend it for a Line. (2D)

class Point2D:
  def __init__(self, x, y):
      self.x = x
      self.y = y
#is a constructor and is automatically called when a new instance of class of creat
  def __str__(self):
      return f"Point at (x = {self.x}, y = {self.y})"
#informal string representation of 2d point and what to do print
  def __repr__(self):
      return f"Point2D({self.x}, {self.y})"
#formal string representation
  def __eq__(self, other):
      return self.x == other.x and self.y == other.y
#defines the = function
  def __add__(self, other):
      return Point2D(self.x + other.x, self.y + other.y)
#defines the + function
  def __sub__(self, other):
      return Point2D(self.x - other.x, self.y - other.y)
#defines the - function
  def __len__(self):
    return 2
#retunrs number of dimension
  def __getitem__(self, key):
    if key == 0:
      return self.x
    elif key == 1:
      return self.y
    else:
      raise IndexError("Point2D only supports index 0 (x) and 1 (y)")
#defines indexing of objects
p1 = Point2D(3, 4)
p2 = Point2D(1, 2)

print(p1)
print(repr(p1))

print(p1 == p2)
print(p1 + p2)
print(p1 - p2)

print(len(p1))
print(p1[0])
print(p1[1])
#print(p1[2])


```

### Univariate Analysis

Univariate analysis focuses on analyzing each feature in the dataset independently.

- **Distribution analysis**: The distribution of each feature is examined to identify its shape, central tendency, and dispersion.

- **Identifying potential issues**: Univariate analysis helps in identifying potential problems with the data such as outliers, skewness, and missing values

#### The shape of a data distribution refers to its overall pattern or form as it is represented on a graph. Some common shapes of data distributions include:

- **Normal Distribution**: A symmetrical and bell-shaped distribution where the mean, median, and mode are equal and the majority of the data falls in the middle of the distribution with gradually decreasing frequencies towards the tails.

- **Skewed Distribution**: A distribution that is not symmetrical, with one tail being longer than the other. It can be either positively skewed (right-skewed) or negatively skewed (left-skewed).

- **Bimodal Distribution**: A distribution with two peaks or modes.

- **Uniform Distribution**: A distribution where all values have an equal chance of occurring.

The shape of the data distribution is important in identifying the presence of outliers, skewness, and the type of statistical tests and models that can be used for further analysis.

#### **Dispersion** is a statistical term used to describe the spread or variability of a set of data. It measures how far the values in a data set are spread out from the central tendency (mean, median, or mode) of the data.

There are several measures of dispersion, including:

- **Range**: The difference between the largest and smallest values in a data set.

- **Variance**: The average of the squared deviations of each value from the mean of the data set.

- **Standard Deviation**: The square root of the variance. It provides a measure of the spread of the data that is in the same units as the original data.

- **Interquartile range (IQR)**: The range between the first quartile (25th percentile) and the third quartile (75th percentile) of the data.

Dispersion helps to describe the spread of the data, which can help to identify the presence of outliers and skewness in the data.

## Why KDE and Box Plot though??

We draw a **boxplot** to quickly see the spread of data, its median, and especially to detect **outliers**, since it neatly summarizes the distribution in a compact form, while we draw a **KDE plot** (Kernel Density Estimate) to understand the **shape of the distribution** smoothly, which helps us see whether the data is normal, skewed, or has multiple peaks.

**In short:**
- **Boxplots** are best for spotting outliers and comparing groups
- **KDE plots** are best for visualizing the overall distribution pattern

1. Introduction

Feature scaling is one of the most essential steps in data preprocessing for machine learning. It ensures that all features in a dataset contribute equally to the learning process, especially in algorithms that calculate distances between data points. Without scaling, models may become biased toward numerically larger values, resulting in inaccurate predictions. While tree-based algorithms are relatively unaffected by feature scaling, most other machine learning and deep learning models benefit greatly from it — both in terms of accuracy and faster convergence during training.

Among various techniques, Normalization and Standardization are the most widely used — and often the most misunderstood. Both fall under the broader category of feature scaling, which involves transforming feature values to a similar scale without distorting their relative differences. This process is particularly important when features have vastly different ranges (e.g., one from 1–10 and another from 1,000–10,000), as it prevents larger-scale features from dominating the model’s learning process. Overall, proper feature scaling not only enhances model performance but also ensures fairness and balance among features during training.

2. Standardization

Standardization, also known as Z-score normalization, is a technique used to transform numerical features so that they have a mean of 0 and a standard deviation of 1.

Standardization is especially useful when features have different units or scales, which can negatively impact certain machine learning algorithms. It is commonly applied in:

Support Vector Machines (SVM)

Linear Regression

Logistic Regression

K-Nearest Neighbors (KNN)

Principal Component Analysis (PCA)

In these algorithms, standardizing the data ensures that no single feature dominates due to its scale. It also improves model performance and speeds up convergence during training.

Although standardization works best when the data is approximately normally distributed, it can still be applied to non-normal data. Unlike normalization, standardization does not restrict data to a fixed range, which makes it less sensitive to outliers while preserving the shape of the distribution.

3. Normalization

Normalization, also called Min-Max Scaling, is a method that transforms feature values to a fixed range, typically between 0 and 1. It is commonly used when the data does not follow a normal distribution and is especially helpful in distance-based algorithms like k-Nearest Neighbors (k-NN) and K-Means Clustering, where differences in feature scales can distort distance calculations.

Normalization works best when there are no extreme outliers, as it is highly sensitive to minimum and maximum values. There are various forms of normalization beyond Min-Max scaling:

Log Normalization: Reduces the impact of large values.

Decimal Scaling: Shifts the decimal point to bring values into a smaller scale.

Mean Normalization: Centers values by subtracting the mean.

These techniques are applied depending on the data distribution and the problem being solved. Overall, normalization brings all feature values onto a common scale, helping models learn more efficiently and fairly.

4. Comparison Between Standardization and Normalization

Although both standardization and normalization are feature scaling techniques, they differ in approach and application. Normalization rescales feature values to a predefined range, typically between 0 and 1, making it suitable for models where feature scales vary widely. Standardization, in contrast, centers data around the mean and scales it according to the standard deviation, producing values with a mean of 0 and a standard deviation of 1.

Normalization is generally more sensitive to outliers, as extreme values can distort the range. Standardization handles outliers better because it accounts for both the mean and variance. In terms of use cases, normalization is often applied in distance-based algorithms like k-NN and K-Means Clustering, where consistent feature scales are crucial. Standardization is essential for gradient-based algorithms such as SVM and is frequently used in dimensionality reduction techniques like PCA, where maintaining correct feature variance is important.
